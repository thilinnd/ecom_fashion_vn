# Gold-label evaluation

This notebook reports the current `set_v2` evaluation on 400 independently reviewed products. It reads frozen evaluation artifacts and does not train models, tune thresholds, or use gold labels for epoch selection.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd()
if not (ROOT / 'fashion_vn').exists():
    ROOT = ROOT.parent
EVAL_DIR = ROOT / 'outputs/evaluation_set_v2'
FIG_DIR = ROOT / 'outputs/figures/evaluation'
summary = json.loads((EVAL_DIR / 'evaluation_summary.json').read_text(encoding='utf-8'))
print(f"Evaluation source: {EVAL_DIR.relative_to(ROOT)}")
print(f"Completed gold records: {summary['completed_gold_rows']:,}")
```

## Evaluation protocol

The fixed score threshold is 0.70 and the runner-up margin is 0.15. Gold validation labels are reserved for reporting and are not used for training, early stopping, calibration, or threshold selection. The main attribute-head comparison is conditional on the category-specific `VISUAL_GROUPS`; end-to-end results remain available in the evaluation summary for pipeline auditing.

```python
scope = pd.DataFrame([
    {'Split': 'Validation', 'Gold records': 200, 'Records in VISUAL_GROUPS': 183, 'Observable groups': 752},
    {'Split': 'Test', 'Gold records': 200, 'Records in VISUAL_GROUPS': 179, 'Observable groups': 747},
])
display(scope)
```

## Overall results in the visual scope

```python
rows = []
for branch_name in ['schema', 'discovered']:
    result = summary['branches'][branch_name]['visual_scope_conditional']
    rows.append({
        'Branch': branch_name.capitalize(),
        'Precision': result['micro']['precision'],
        'Recall': result['micro']['recall'],
        'Micro-F1': result['micro']['f1'],
        'Macro-F1': result['macro_f1'],
        'Macro-mAP': result['macro_map'],
        'Coverage': result['group_coverage'],
        'Selective exact': result['selective_group_exact_match_accuracy'],
        'Abstention': result['group_abstention_rate'],
    })
overall = pd.DataFrame(rows)
display(overall.style.format({column: '{:.4f}' for column in overall.columns if column != 'Branch'}))
```

## Schema validation and test results

```python
schema_splits = []
diagnostics = summary['branches']['schema']['visual_scope_subgroup_diagnostics']
for split_key, split_label in [('split_val', 'Validation'), ('split_test', 'Test')]:
    result = diagnostics[split_key]
    schema_splits.append({
        'Split': split_label,
        'Number of records': result['evaluated_products'],
        'Number of groups': result['evaluated_group_instances'],
        'Precision': result['micro']['precision'],
        'Recall': result['micro']['recall'],
        'Micro-F1': result['micro']['f1'],
        'Macro-F1': result['macro_f1'],
        'Macro-mAP': result['macro_map'],
        'Coverage': result['group_coverage'],
        'Selective exact': result['selective_group_exact_match_accuracy'],
        'Abstention': result['group_abstention_rate'],
    })
schema_split_table = pd.DataFrame(schema_splits).set_index('Split').T
display(schema_split_table.style.format('{:.4f}', subset=pd.IndexSlice[schema_split_table.index[2:], :]))
```

## Results by attribute group

```python
schema_group = pd.read_csv(EVAL_DIR / 'schema_visual_scope_per_group.csv')
discovered_group = pd.read_csv(EVAL_DIR / 'discovered_visual_scope_per_group.csv')
group_table = schema_group[['group', 'group_instances', 'f1', 'coverage', 'exact_match_accuracy', 'selective_exact_match_accuracy']].rename(columns={
    'group': 'Group', 'group_instances': 'Cases', 'f1': 'Schema F1',
    'coverage': 'Schema coverage', 'exact_match_accuracy': 'Schema exact match',
    'selective_exact_match_accuracy': 'Schema selective exact',
})
group_table = group_table.merge(
    discovered_group[['group', 'f1']].rename(columns={'group': 'Group', 'f1': 'Discovered F1'}),
    on='Group', how='left',
)
order = ['color', 'sleeve', 'neckline', 'pattern', 'fit', 'length', 'pant_cut', 'waist']
group_table['Group'] = pd.Categorical(group_table['Group'], categories=order, ordered=True)
group_table = group_table.sort_values('Group').reset_index(drop=True)
display(group_table.style.format({column: '{:.4f}' for column in group_table.columns if column not in ['Group', 'Cases']}))
```

## Category and attribute confusions

The category table compares `category_gold` with the title-derived `category_hint`; it is not an image-category classifier. Attribute confusion tables exclude abstentions as label-to-label confusion pairs, while abstention remains reflected in recall, coverage, and F1.

```python
category_long = pd.read_csv(EVAL_DIR / 'category_confusion.csv')
category_matrix = category_long.pivot_table(index='category_gold', columns='category_hint', values='count', aggfunc='sum', fill_value=0)
display(category_matrix)

schema_confusions = pd.read_csv(EVAL_DIR / 'schema_visual_scope_attribute_confusions.csv')
discovered_confusions = pd.read_csv(EVAL_DIR / 'discovered_visual_scope_attribute_confusions.csv')
print('Most frequent schema attribute confusions')
display(schema_confusions.head(15))
print('Discovered-branch attribute confusions')
display(discovered_confusions.head(15))
```

## Interpretation

The schema branch reaches a visual-scope micro-F1 of 0.3203 with precision 0.7179, while its 0.6977 abstention rate limits recall to 0.2062. Validation and test micro-F1 remain nearly identical (0.3197 and 0.3210), although test Macro-F1 is lower, indicating uneven performance across attribute values. The discovered branch should be interpreted primarily as a knowledge-discovery mechanism: its concepts can expand vocabulary and suggest new taxonomy structure, while direct mapping to the current gold taxonomy remains limited (micro-F1 0.0527 and abstention 0.9644). The improvement over the previous report is mainly attributable to evaluating set_type as the parent label set.
